# คุณวิเคราะห์ข้อมูลเชิงลึกของ State.csv ได้อย่างไรบ้าง

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot
from pathlib import Path

state = pd.read_csv(Path("state.csv"))
```

"State","Population","Murder Rate","Abbreviation"

```python
state.groupby('State')['Population'].mean().sort_values(ascending=False)
```

```python
new_state = state[::]
new_state['Population'] = new_state['Population'] / 1000000
```

```python
new_state.groupby('State')[['Population', 'Murder Rate']].mean().sort_values(ascending=False, axis=0, by='Murder Rate')
```

```python
x = new_state['Population']
y = new_state['Murder Rate']

pyplot.scatter(x, y)
pyplot.xlabel("Population(m)")
pyplot.ylabel("Murder Rate")
pyplot.show()
```
