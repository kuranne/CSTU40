### นักศึกษาต้องระมัดระวังว่า ค่าของ RGB ต้องไม่เกิน 255
> Hint: ศึกษาการใช้งาน ``np.clip()``

### <u>หมายเหตุ</u> ไม่อนุญาตให้นักศึกษาใช้ iterative instructions (while, for) ในการทำงาน




````python
import numpy as np
import matplotlib.pyplot as plt

image = np.array([
    [[255,   0,   0], [200,  50,  50], [150, 100,  50], [100, 150,  50], [ 50, 200,  50]],
    [[  0, 255,   0], [ 50, 200,  50], [100, 150, 100], [150, 100, 100], [200,  50, 100]],
    [[  0,   0, 255], [ 50,  50, 200], [100, 100, 150], [150, 150, 100], [200, 200,  50]],
    [[255, 255,   0], [200, 200,  50], [150, 150, 100], [100, 100, 150], [ 50,  50, 200]]
], dtype=np.uint8)


# Q.4 your code starts from here.
# Write down your answer after the question number below.

def show_img(arr):
    plt.imshow(arr)
    plt.axis(False)
    plt.show()

# 4.1.
print(image.shape, image.ndim, image.size, image.dtype)

# 4.2.
print(image.shape[0] * image.shape[1])

# 4.3.
print(image[0, 1])

# 4.4.
red = image[:,:,0].copy()
print(red, "\n")

# 4.5.
print(f"""
Average Red     = {np.mean(image[:,:,0])}
Average Green   = {np.mean(image[:,:,1])}
Average Blue    = {np.mean(image[:,:,2])}
    """
)

# 4.6.

no_red = image.copy()
no_red[:,:,0] = 0

show_img(no_red)
show_img(image)

# 4.7.
bright_image = np.array(image, dtype=np.uint16)
bright_image += 50
bright_image = np.where(bright_image > 255, 255, bright_image)
bright_image = np.array(bright_image, dtype=np.uint8)

show_img(bright_image)

#plt.imshow(image)
#plt.axis("off")
#plt.show()
````
