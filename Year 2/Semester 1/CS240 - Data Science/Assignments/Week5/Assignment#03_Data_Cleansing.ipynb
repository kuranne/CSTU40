##### 4.1.4. (7 คะแนน) Fill missing values in every column (except the 'country' and 'children' column) with ``0``.
Show how you verify that all NaN values in the specified columns have been replaced.

Students must provide comments along your code to explain your solution.  

[use the following code area for answering this question]

````python
# Q.4.1.4.

# init target_cols to select all columns then
# filter country and children col out of target_cols
target_cols = dataframe.columns.difference(["country", "children"])

# replace all null value in target_cols with 0 by `.fillna(0)`
dataframe[target_cols] = dataframe[target_cols].fillna(0)

# check the result with `dataframe.isnull().sum().sum()
# if result is equal to the sum of na in country + children
# determine that all of null value is replaced with 0, except for in country and children columns
print("Country's na + Children's na: ", dataframe[['country', 'children']].isnull().sum().sum())
print("Total of data frame's na: ", dataframe.isnull().sum().sum())
````

##### 4.1.5. (6 คะแนน) Fill all missing values in column 'country' with the most-frequently occurred (``mode``) value of the column.

Show how you verify that all NaN values in the column 'country' have been replaced.

Students must provide comments along your code to explain your solution.

[use the following code area for answering this question]


````python
# Q.4.1.5.
print("Before filling, count of country's na:", dataframe['country'].isnull().sum())
# to proof all of country's na value would replaced with mode, after filling, the count of country's na would be 0

# bring all mode of country, select for 1st mode index
# then fill all na value under country column with 1st mode value
country_mode = dataframe['country'].mode()[0]
# print(country_mode)
dataframe['country'] = dataframe['country'].fillna(country_mode)

print("After filling, count of country's na:", dataframe['country'].isnull().sum())
# the after filling result should be 0, to verify the missing data under country column is replaced
````

#####4.1.6. (6 คะแนน) Fill all missing values in column 'children' with the rounded ``mean`` value of the column.

````python
# Q.4.1.6.
# dataframe['children'].value_counts()
print("Before filling, children's na: ", dataframe['children'].isnull().sum())

children_mean = dataframe['children'].mean()
dataframe['children'] = dataframe['children'].fillna(children_mean.round())

print("After filling, children's na: ", dataframe['children'].isnull().sum())
````

###5. (20 คะแนน) Data Validity

There are many rows in the data set that have zero guests including adults, children and babies. These type of rows does not make any sense.

Find booking rows with total guests equal to zero and remove all of them.

#####5.1.(5 คะแนน) Identify how many rows in the data set that have zero guests including adults, children and babies.

```
Expected Output:
Number of rows with total 0 guests = ..........................
```

````python
#Q.5.1.

guest = ['adults', 'children', 'babies']
zero_guest = dataframe[dataframe[guest].sum(axis=1) == 0]
print("Number of rows with total 0 guests =", len(zero_guest))
````

#####5.3.(10 คะแนน) Drop all rows in the data set that have zero guests including adults, children and babies.

Verify also that all rows with total 0 guests have been removed.

Students must provide comments along your code to explain your solution.  

<u>Hint</u>: ใช้ index ในการระบุแถวข้อมูลที่ต้องการ drop

[use the following code area for answering this question]

````python
#Q.5.3.
zero_guest = dataframe[dataframe[guest].sum(axis=1) == 0]
print("Before dropping, number of 0 guest rows: ", len(zero_guest))

dataframe = dataframe.drop(zero_guest.index)

zero_guest = dataframe[dataframe[guest].sum(axis=1) == 0]
print("After dropping, number of 0 guest rows: ", len(zero_guest))
# to verify that all rows with total 0 guests have been removed,
# the retult of number of 0 guest rows after dropped must be 0
````

###6. (23 คะแนน) Data Validity (More -- Challenge)

There might be some rows in the data set that there are children or baby guests without adults. These type of rows does not make any sense.

Find booking rows with such condition and remove all of them.

#####6.1.(18 คะแนน) Identify how many rows in the data set that have children or babies guests, but zero adults.

```
Expected Output:
Number of rows with children or babies guests, but zero adults = .............
```

````python
#Q.6.1.
no_adult_rows = dataframe[dataframe['adults'] == 0]
no_adult_but_have_minor = no_adult_rows[ no_adult_rows[['children', 'babies']].sum(axis=1) > 0 ]
print("Number of rows with children or babies guests, but zero adults =", len(no_adult_but_have_minor))
````

#####6.2.(5 คะแนน) Drop all rows in the data set that have children or babies guests, but zero adults.

Verify also that all rows with total 0 guests have been removed.

Students must provide comments along your code to explain your solution.  

````python
#Q.6.2.
no_adult_rows = dataframe[dataframe['adults'] == 0]
no_adult_but_have_minor = no_adult_rows[ no_adult_rows[['children', 'babies']].sum(axis=1) > 0 ]
print("Before dropping, number of zero adult but have atlest 1 minors =", len(no_adult_but_have_minor))

dataframe = dataframe.drop(no_adult_but_have_minor.index)

no_adult_rows = dataframe[dataframe['adults'] == 0]
no_adult_but_have_minor = no_adult_rows[ no_adult_rows[['children', 'babies']].sum(axis=1) > 0 ]
print("After dropping, number of zero adult but have atlest 1 minors =", len(no_adult_but_have_minor))
# to verify all if no adult but have children or babies rows has been dropped
# the after dropping result must equal 0
````
